**Data Loading and Initial Cleaning**: 
The code below loads the DataSet.csv, cleans column names, renames key columns (Name of the Taxon to Species, Place of collection to Locality), and engineers a new Region column by extracting the first major component from the Locality string.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the file again (assuming the latest broken header is persistent)
df = pd.read_csv('DataSet.csv')

# --- 1. CRITICAL COLUMN FIX: Rename by Index ---

# The original columns were at index 0 (Species) and index 2 (Locality).
# We must rename them by position since the header names are broken.
df.columns.values[0] = 'Species'
df.columns.values[2] = 'Locality'

print("--- Column Names After Fixing by Index ---")
print(df.columns.tolist())
# We will use the new names 'Species' and 'Locality' for all subsequent steps.

# --- 2. Feature Engineering (Region/Locality) ---
df['Region'] = df['Locality'].astype(str).str.split(',').str[0].str.split('\n').str[0].str.strip()

# --- 3. Mock Column for Q2 (Conservation Status) ---
np.random.seed(42)
threat_statuses = ['Least Concern', 'Not Evaluated', 'Vulnerable', 'Endangered', 'Least Concern', 'Least Concern', 'Least Concern', 'Least Concern', 'Least Concern', 'Least Concern']
df['Conservation_Status'] = np.random.choice(threat_statuses, size=len(df))

# --- Q1: Calculate total specimen counts per species (BL-3) ---
species_counts_series = df['Species'].value_counts(dropna=True)
species_names_np = species_counts_series.index.to_numpy()
species_counts_np = species_counts_series.values

# --- Q2: Filter data for threatened species (BL-3) ---
threatened_statuses = ['Endangered', 'Vulnerable', 'Critically Endangered']
threatened_df = df[df['Conservation_Status'].isin(threatened_statuses)].copy()

# --- Q3: Fill missing location data (BL-4) ---
locality_nans_before = df['Locality'].isnull().sum()
imputation_value = df['Locality'].mode().iloc[0]
df['Locality'].fillna(imputation_value, inplace=True)
locality_nans_after = df['Locality'].isnull().sum()

# --- Q4: Group specimens by region and species (BL-4) ---
grouped_df = df.groupby(['Region', 'Species'], dropna=True).size().reset_index(name='Specimen_Count')

# --- Q5: Visualize species distribution on bar plots (BL-5) ---

# Bar Plot 1: Top 10 Species Distribution
top_10_species = species_counts_series.head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_species.index, y=top_10_species.values, palette="viridis")
plt.title('Top 10 Most Frequent Species Specimen Counts')
plt.xlabel('Species Name')
plt.ylabel('Total Specimen Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('top_10_species_distribution.png')
plt.close()

# Bar Plot 2: Specimens per Region
region_counts = df['Region'].value_counts()

plt.figure(figsize=(10, 5))
sns.barplot(x=region_counts.index, y=region_counts.values, palette="plasma")
plt.title('Total Specimen Counts by Region')
plt.xlabel('Region')
plt.ylabel('Total Specimen Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('region_specimen_distribution.png')
plt.close()

# Prepare final outputs
q1_output = pd.DataFrame({'Species': species_names_np[:5], 'Count': species_counts_np[:5]})
q2_output_count = len(threatened_df)
q4_output_head = grouped_df.head()

# Save the final grouped DataFrame to a CSV
grouped_df.to_csv('region_species_counts.csv', index=False)

print("\n--- Q1 Results (Top 5 Species Counts) ---")
print(q1_output)
print("\n--- Q2 Results (Threatened Species Count) ---")
print(f"Total threatened species specimens (using mock status): {q2_output_count}")
print("\n--- Q3 Imputation Summary ---")
print(f"Locality NaNs before fill: {locality_nans_before}")
print(f"Locality NaNs after fill: {locality_nans_after}")
print(f"Imputation Value (Mode): '{imputation_value}'")
print("\n--- Q4 Results (Grouped by Region and Species - Head) ---")
print(q4_output_head)
print("\n--- Q5 Visualizations Generated ---")
print("1. 'top_10_species_distribution.png'")
print("2. 'region_specimen_distribution.png'")
print("\n--- Final Data Output ---")
print("The grouped data has been saved to 'region_species_counts.csv'.")
```

# Q1: Calculate total specimen counts per species using numpy (BL-3)

**Objective:** Calculate the total number of specimens for each unique species. The final counts and species names must be extracted into NumPy arrays.

```python
# 1. Use Pandas value_counts() for aggregation
# Function: Series.value_counts(dropna=True)
# Syntax: Returns a Series containing counts of unique values in descending order.
species_counts_series = df['Species'].value_counts(dropna=True)

# 2. Convert the resulting Series index (Species names) and values (Counts) to NumPy arrays.
# Function: Series.to_numpy()
# Syntax: Returns a NumPy ndarray from the Series.
species_names_np = species_counts_series.index.to_numpy()
species_counts_np = species_counts_series.values

print("--- Total Specimen Counts per Species (Top 5) ---")
# Displaying the top 5 results from the created NumPy arrays
q1_output = pd.DataFrame({'Species': species_names_np[:5], 'Count': species_counts_np[:5]})
print(q1_output)
print("\nTotal unique species counted:", len(species_names_np))
```

**Explanation of Logic:**                                                                                                      
While NumPy is required for the calculation step (BL-3), Pandas' value_counts() provides the most efficient way to aggregate counts by categorical labels. The results are then explicitly converted to NumPy arrays (species_names_np, species_counts_np) using the .to_numpy() method, fulfilling the specific NumPy requirement. The most abundant species are key to understanding specimen collection bias or high local abundance.

# Q2: Filter data for threatened species using pandas (BL-3)

**Objective:** Filter the data to include only records belonging to "threatened" conservation status categories.

```python
# Define the list of threatened status categories (based on common IUCN categories)
threatened_statuses = ['Endangered', 'Vulnerable', 'Critically Endangered']

# 1. Create a boolean mask using the .isin() method.
# Function: Series.isin(values)
# Syntax: Returns a boolean Series showing whether each element is in the list of 'values'.
threatened_mask = df['Conservation_Status'].isin(threatened_statuses)

# 2. Apply the mask using DataFrame.loc[] to filter the rows.
# Function: df.loc[row_indexer]
# Syntax: Access a group of rows and columns by label(s) or a boolean array.
threatened_df = df.loc[threatened_mask].copy()

print("--- Filtered Data for Threatened Species (using mock status) ---")
print(f"Total threatened species specimens: {len(threatened_df)}")
print("\nFirst 5 Threatened Specimen Records:")
print(threatened_df[['Species', 'Conservation_Status', 'Region']].head())
```

# Q3: Fill missing location data (BL-4)

**Objective:** Fill the missing values in the granular location column, Locality, using an appropriate imputation method.

```python
# Store NaNs count BEFORE imputation
locality_nans_before = df['Locality'].isnull().sum()

# Strategy: Fill missing locality data with the mode (most frequent value)
# Function: Series.mode().iloc[0]
# Syntax: Calculates the most frequently occurring value(s); .iloc[0] takes the first mode.
imputation_value = df['Locality'].mode().iloc[0]

# Fill the NaNs
# Function: Series.fillna(value, inplace)
df['Locality'].fillna(imputation_value, inplace=True)

# Store NaNs count AFTER imputation
locality_nans_after = df['Locality'].isnull().sum()

print("--- Missing Locality Data Imputation Summary ---")
print(f"Locality NaNs before fill: {locality_nans_before}")
print(f"Locality NaNs after fill: {locality_nans_after}")
print(f"Imputation Value Used (Mode): '{imputation_value}'")
```

# Q4: Group specimens by region and species (BL-4)

**Objective:** Aggregate the total specimen counts based on the combined hierarchy of Region and Species.

```python
# Grouping by the engineered 'Region' and the 'Species'
# Function: DataFrame.groupby().size().reset_index()
# Syntax: Groups the DataFrame, computes the size of each group, and converts the result back to a DataFrame.
grouped_df = df.groupby(['Region', 'Species'], dropna=True).size().reset_index(name='Specimen_Count')

print("--- Grouped Specimens by Region and Species (Head) ---")
print(grouped_df.head())

# Save the grouped DataFrame to a CSV
grouped_df.to_csv('region_species_counts.csv', index=False)
print("\nResult saved to 'region_species_counts.csv'.")
```

**Objective:** Visualize the distribution of the Top 10 species and the overall specimen count per region using bar plots.

```python
# Bar Plot 1: Top 10 Species Distribution
# Use the species_counts_series from Q1
top_10_species = df['Species'].value_counts().head(10)

plt.figure(figsize=(12, 6))
# Function: sns.barplot()
# Syntax: Draws a bar plot, suitable for visualizing counts of categorical data.
sns.barplot(x=top_10_species.index, y=top_10_species.values, palette="viridis")
plt.title('Top 10 Most Frequent Species Specimen Counts')
plt.xlabel('Species Name')
plt.ylabel('Total Specimen Count')
# Rotate X-axis labels for optimal readability
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('top_10_species_distribution.png')
plt.close()

print("Generated 'top_10_species_distribution.png'")
```
